```python
data = ["Cat", 123, 9.8]
series = pd.Series(data)
print(series)
```

```python
data = [100, 123, 333, 500, 111, 300]
series = pd.Series(data, index= ["first", "second", "third", "fourth", "fifth", "sixth"])
print(series.loc["first"])
print(series.iloc[0])
print(series[series < 200])
```

```python
calories = {"Day 1" : 1750, "Day 2" : 2100, "Day 3" : 1700, "Day 4" : 1900}
series = pd.Series(calories)
series.loc["Day 3"] += 1200
print(series[series > 1800])
```

```python
data = {"Name" : ["Luffy", "Zorro", "Sanji"],
        "Age" : [19, 20, 13]
}
df = pd.DataFrame(data, index=["captain", "vice-captain", "cook"])
print(df.loc['captain'])
print(df.iloc[1])
```

```python
# Add a new column 
df['bounty'] = ['3 Billion', '3 Billion', '2 Billion']

# Add a new row
new_row = pd.DataFrame([{"Name": "Chopper", "Age" : 9, "bounty" : "10000"},{"Name": "Nami", "Age" : 16, "bounty" : "1 Billion"}], index = ["doctor","navigator"])

df = pd.concat([df, new_row])

print(df)
```

```python
df = pd.read_csv("./datasets/one_piece_100_plus.csv")
print(df.sum(numeric_only=True))
print(df.count())
```

```python
# Data cleaning
# Drop irrelevant cols
df = df.drop(columns=["gender"])
print(df)
```

```python
#handle missing data
# df = df.dropna(subset=["fruit_type"])
# print(df)

# replace value
df = df.fillna({"devil_fruit": "Unknown"})
print(df)
```

```python
#fix inconsistent value
df["fruit_type"] = df["fruit_type"].replace({"Paramecia": "PARAMECIA"})
print(df)
```

```python
# standardize text
df["name"] = df["name"].str.lower()
print(df)
```

```python
#remove duplicate
df = df.drop_duplicates()
print(df)
```
