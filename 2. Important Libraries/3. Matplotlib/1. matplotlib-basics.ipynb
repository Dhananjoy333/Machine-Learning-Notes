```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([25, 18, 36, 13])
y3 = np.array([17, 27, 21, 19])

line_styles = dict(marker="o", ms =20, mfc="red",mec="black", linestyle = "solid", linewidth= 3)

plt.plot(x, y1, color = "#1c5bfc", **line_styles)
plt.plot(x, y2, color = "green",**line_styles)
plt.plot(x, y3, color= "orange",**line_styles)

plt.show()
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([25, 18, 36, 13])
y3 = np.array([17, 27, 21, 19])

plt.title("Class size", fontsize = 20, family = "Arial", fontweight = "bold", color = "blue")

plt.xlabel("Year", fontsize = 15, family = "Arial", fontweight = "bold", color = "green")

plt.ylabel("Students", fontsize = 15, family = "Arial", fontweight = "bold", color = "purple")

plt.tick_params(axis="both", colors="cyan")

plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)

plt.xticks(x)

plt.show()
```

```python
x = [1,2,3,4,5]
y = [5,10,15,20,25]

plt.grid(axis="y", linewidth = 3, color = "lightgray", linestyle = "dashed")

plt.plot(x,y)
plt.show()
```

```python
categories = np.array(["Grains", "Fruit", "Vegetables", "Proteins", "Dairy", "Sweets"])
values = np.array([4, 3, 2, 5, 3, 1])

plt.bar(categories, values, color = "cyan")
# plt.barh(categories, values, color = "cyan")

plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")

plt.show()
```

```python
categories = np.array(["Freshmen", "Sophomores", "Juniors", "Seniors"])
values = np.array([300, 250, 275, 225])
colors = ["red", "yellow", "blue", "green"]

plt.pie(values, labels=categories, autopct="%1.1f%%", colors=colors, explode=[0,0,0,0.1], shadow=True, startangle=90)

plt.title("College")

plt.show()
```

```python
x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8]) #hours_studies
y1 = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) # marks

x2 = np.array([0, 1, 2, 2, 3, 3, 5, 6, 7, 8, 8]) #hours_studies
y2 = np.array([50, 58, 67, 71, 63, 73, 75, 79, 92, 80, 97]) # marks

plt.scatter(x1,y1, color= "skyblue", alpha= 0.5, s = 100, label = "Class A")
plt.scatter(x2,y2, color= "red", alpha= 0.5, s = 100, label="Class B")

plt.title("Test Scores")
plt.xlabel("Hours Studies")
plt.ylabel("Grade")

plt.legend()
plt.show()
```

```python
scores = np.random.normal(loc= 80, scale= 10, size= 100)
scores = np.clip(scores, 0, 100)

plt.hist(scores, bins= 10, color= "lightgreen", edgecolor= "black")

plt.title("Exam scores")
plt.xlabel("Score")
plt.ylabel("Number of students")

plt.show()
```

```python
#subplots
x = np.array([1,2,3,4,5])

figure, axes = plt.subplots(2, 2)

axes[0,0].plot(x,x*2, color= "red")
axes[0,0].set_title("x*2")

axes[0,1].plot(x, x**2, color="blue")
axes[0,1].set_title("x**2")

axes[1,0].plot(x, x**3, color= "green")
axes[1,0].set_title("x**3")

axes[1,1].plot(x, x**4, color= "yellow")
axes[1,1].set_title("x**4")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("./datasets/one_piece_100_plus.csv")

df = df.drop_duplicates()

faction_count = df["faction"].value_counts(ascending=True)

plt.barh(faction_count.index, faction_count.values, color = "cyan", edgecolor= "black")

plt.title("Total character in each faction")
plt.xlabel("Count",color = "red")
plt.ylabel("Faction", color = "red")

plt.tight_layout()

plt.show()
```

```python
grouped = df.groupby(["faction", "gender"])["bounty"].mean()
grouped.unstack().plot(kind="bar")
plt.show()
```
