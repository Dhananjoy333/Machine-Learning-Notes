```python
import numpy as np
```

```python
A = np.array([
        [-1, 3],
        [3, 2]
    ], dtype=np.dtype(float))

b = np.array([7, 1], dtype=np.dtype(float))

print("Matrix A:")
print(A.shape)
print("\nArray b:")
print(b.shape)

inverse = np.linalg.inv(A)
sol = inverse @ b
solve = np.linalg.solve(A,b)
print(f'solution: \n {sol}')
print(f'solution: \n {solve}')
```

```python
A = np.array([
        [-1, 3],
        [3, 2]
    ], dtype=np.dtype(float))
det = np.linalg.det(A)
print(det)

rank = np.linalg.matrix_rank(A)
print(rank)

eig = np.linalg.eig(A)
print(eig)
eigenvalues, eigenvectors = np.linalg.eig(A)
print(eigenvalues)

eigh = np.linalg.eigh(A)
print(eigh)

svd = np.linalg.svd(A)
print(svd)

norm = np.linalg.norm(A)
print(norm)
```
