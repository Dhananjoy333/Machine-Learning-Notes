# Principal Component Analysis
PCA is a unsupervised dimensional reduction technique where we create principle components. 
In detail explanation : [PCA](../../../../1.%20Math%20for%20Machine%20Learning/1.1%20Computational%20Linear%20Algebra/5.%20Dimensional%20Reduction%20and%20PCA.md). In PCA we project the datapoints into a single line decreasing the number of dimensions. We choose a projection line where datapoints have max variance.

Steps to dimensional reduction :
1. Center the dataset ($C_d$). 
2. Find covariance ($Cov_d$) matrix of the $C_d$. For a dataset of $n*m$ size ($n$ = no. of datas, $m$ = no. of features), the $Cov_d$ matrix will be of size $m*m$.
3. Find eigenvalues and eigenvectors of $Cov_d$. For a covariance matrix of size $m*m$, we will get $m$ no. of eigenvalues and their corresponding eigenvectors.
4. Sort the eigenvalues in descending order.
5. Select the no. of eigenvalues to which we want the dimension to be reduced to from size $m$. If we select 1 eigenvalue we get 1 vector i.e., eigenvector corresponding to the selected eigenvalue. If we select first $n$ ($n<m$) no. of eigenvalues we get a matrix of eigenvectors of those selected eigenvalues.
6. Now we project the datapoints on the vector/matrix of eigenvectors, we know we can project one vector to another by dot product. Hence, we dot point matrix of datapoints and vector/matrix of eigenvectors.  
Hence, our dimension has been reduced from $m$ to $n$($n<m$). 

```python
import numpy as np
import pandas as pd
```

```python
# Dataset has 7 columns/features/dimensions.
df = pd.read_csv('./datasets/students.csv')
centered_df = df.copy()
means = df.mean()

# Step-1 : centering the dataset.
centered_df = df - means

# Step-2 : finding covariance of centered dataset.
covariance_mat = centered_df.cov()

# Step-3 : finding eigenvalues and eigenvectors of covariance matrix
eigenvalues, eigenvectors = np.linalg.eig(covariance_mat)

# Step-4 : sorting eigenvalues in descending order. Here we not rearranging the actual eigenvalue places instead we simply get the index of eigenvalues in descending order, [0 1 2 5 6 4 3]. We are directly not rearranging because eigenvalues and eigenvectors are connected simply rearranging the eigenvalues will cause mismatch between eigenvalues and eigenvectors.
idx = np.argsort(eigenvalues)[::-1]

# rearrange eigenvalues based on descending index.
eigenvalues = eigenvalues[idx]

# rearrange eigenvectors based on index of eigenvalues (descending index).
eigenvectors = eigenvectors[:, idx]

# Step-5 : Choose the no. of eigenvectors upto which we want to reduce the dimension, we chose 5 i.e., dimension 7 --> dimension 5.
top5 = eigenvectors[:, :5]

# Step-6 : Prohject the data by dot product.
reduced_df = centered_df @ top5

print(reduced_df)
```
